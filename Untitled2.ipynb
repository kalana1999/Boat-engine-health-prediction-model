## SECTION 1: Setup and Imports

```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix, precision_score, recall_score, f1_score
import joblib

np.random.seed(42)
RANDOM_STATE = 42

print("✅ load all libraries are successfully !")
```

## SECTION 2: Data Loading and Initial Inspection

```python

df = pd.read_csv("/content/final_sensor_data_for_ml.csv")

print("Dataset shape:", df.shape)
display(df.head())
```

## SECTION 3: Dataset Structure and Summary Statistics

```python

df.info()
display(df.describe(include="all").T)
```

## SECTION 4: Exploratory Data Analysis (EDA)

```python

plt.figure(figsize=(8, 4))
sns.countplot(data=df, x='Fault_Label', palette='Set2')
plt.title("Distribution of Fault Labels", fontsize=14)
plt.xlabel("Fault Label")
plt.ylabel("Count")
plt.show()

# Correlation Heatmap
plt.figure(figsize=(8, 6))
numeric_df = df[['Temperature', 'Smoke', 'Flame', 'Vibration', 'Fault_Label']]
sns.heatmap(numeric_df.corr(), annot=True, cmap='Blues', fmt='.2f')
plt.title("Sensor Correlation Heatmap", fontsize=14)
plt.show()
```

## SECTION 5: Data Preparation & Splitting

```python

X = df[['Temperature', 'Smoke', 'Flame', 'Vibration']]
y = df['Fault_Label']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=RANDOM_STATE)

print(f"Training Data Size: {X_train.shape[0]} rows")
print(f"Testing Data Size: {X_test.shape[0]} rows")
```

## SECTION 6: Logistic Regression Baseline Model

```python

lr_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median", add_indicator=True)),
    ("scaler", StandardScaler()),
    ("classifier", LogisticRegression(max_iter=2000, class_weight="balanced", random_state=RANDOM_STATE))
])

lr_pipeline.fit(X_train, y_train)
lr_predictions = lr_pipeline.predict(X_test)

print("📉 Logistic Regression Validation Results:\n")
print(classification_report(y_test, lr_predictions, zero_division=0))
print("Confusion Matrix:\n", confusion_matrix(y_test, lr_predictions))
```

## SECTION 7: Random Forest Classifier

```python

rf_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median", add_indicator=True)),
    ("classifier", RandomForestClassifier(
        n_estimators=300,
        max_depth=10,
        min_samples_leaf=3,
        class_weight="balanced_subsample",
        random_state=RANDOM_STATE,
        n_jobs=-1
    ))
])

rf_pipeline.fit(X_train, y_train)
rf_predictions = rf_pipeline.predict(X_test)

print("🚀 Random Forest Validation Results:\n")
print(classification_report(y_test, rf_predictions, zero_division=0))
print("Confusion Matrix:\n", confusion_matrix(y_test, rf_predictions))
```

## SECTION 8: Baseline and Random Forest Model Comparison

```python

model_predictions = {
    "Logistic Regression": lr_predictions,
    "Random Forest": rf_predictions
}

comparison_results = []
for model_name, predictions in model_predictions.items():
    comparison_results.append({
        "Model": model_name,
        "Accuracy": accuracy_score(y_test, predictions),
        "Precision": precision_score(y_test, predictions, average="macro", zero_division=0),
        "Recall": recall_score(y_test, predictions, average="macro", zero_division=0),
        "F1-Score": f1_score(y_test, predictions, average="macro", zero_division=0)
    })

model_comparison = pd.DataFrame(comparison_results)
print("📊 Model Comparison Table:")
display(model_comparison.round(4))
```

## SECTION 9: Save the Best Model (.pkl)

```python

model_filename = 'seaguardian_ai_model_updated.pkl'
joblib.dump(rf_pipeline, model_filename)

print(f"💾 Best Model  '{model_filename}'  Save the successfully!")
```
